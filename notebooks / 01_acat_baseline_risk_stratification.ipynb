**Build the Leakage-Free Preprocessing & Baseline Script**

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import GroupShuffleSplit
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import precision_recall_curve, auc, brier_score_loss

print("Loading dataset from local folder...")
df = pd.read_csv('data/sample_condition_occurrence.csv')

print("Engineering features and target...")
df['condition_start_date'] = pd.to_datetime(df['condition_start_date'])

# Aggregate features per person_id
patient_agg = df.groupby('person_id').agg(
    total_conditions=('condition_concept_id', 'count'),
    distinct_conditions=('condition_concept_id', 'nunique'),
    first_condition_date=('condition_start_date', 'min'),
    last_condition_date=('condition_start_date', 'max')
).reset_index()

# Calculate observation span in days
patient_agg['obs_span_days'] = (patient_agg['last_condition_date'] - patient_agg['first_condition_date']).dt.days

# Define target: High-utilization patient (top 20% by total conditions)
threshold = patient_agg['total_conditions'].quantile(0.80)
patient_agg['target'] = (patient_agg['total_conditions'] > threshold).astype(int)

# Features (X) and Target (y)
X = patient_agg[['distinct_conditions', 'obs_span_days']]
y = patient_agg['target']
groups = patient_agg['person_id'] # Crucial for patient-level splitting!

print("Performing patient-level train/test split (preventing data leakage)...")
gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
train_idx, test_idx = next(gss.split(X, y, groups))

X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

print(f"Training set patients: {len(X_train)}, Testing set patients: {len(X_test)}")

print("Training baseline logistic regression model...")
baseline_model = LogisticRegression(class_weight='balanced', random_state=42)
baseline_model.fit(X_train, y_train)

# Evaluate metrics
y_pred_proba = baseline_model.predict_proba(X_test)[:, 1]
precision, recall, _ = precision_recall_curve(y_test, y_pred_proba)
pr_auc = auc(recall, precision)
brier = brier_score_loss(y_test, y_pred_proba)

print("\n--- Baseline Evaluation Results ---")
print(f"PR-AUC: {pr_auc:.4f}")
print(f"Brier Score (Calibration): {brier:.4f}")
```

**This is a case of Target Leakage / Trivial Proxy Target:**

Look at how the target was defined in the code: patient_agg['target'] = (patient_agg['total_conditions'] > threshold).astype(int).

Look at the features (X): ['distinct_conditions', 'obs_span_days'].

The Catch: distinct_conditions and total_conditions are mathematically almost identical.

Instead of predicting **a self-referential count** (which causes leakage), a real clinical prediction model looks backward in time to predict what will happen forward in time.

**We will split each patient's medical history into two windows:**

**Observation Window (History):** Features built from their first year of data.

**Prediction Window (Future Outcome):** Whether they return with a high utilization event or specific condition in the subsequent period.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import GroupShuffleSplit
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import precision_recall_curve, auc, brier_score_loss

print("Loading dataset...")
df = pd.read_csv('data/sample_condition_occurrence.csv')
df['condition_start_date'] = pd.to_datetime(df['condition_start_date'])

# 1. Establish a Temporal Cutoff per Patient (e.g., split their history in half)
print("Setting up temporal observation and prediction windows...")
patient_first_last = df.groupby('person_id')['condition_start_date'].agg(['min', 'max']).reset_index()
patient_first_last['median_date'] = patient_first_last['min'] + (patient_first_last['max'] - patient_first_last['min']) / 2

# Merge median date back to main dataframe
df = df.merge(patient_first_last[['person_id', 'median_date']], on='person_id', how='left')

# 2. Observation Window (Features): Events BEFORE the median date
obs_df = df[df['condition_start_date'] <= df['median_date']]
X_features = obs_df.groupby('person_id').agg(
    hist_distinct_conditions=('condition_concept_id', 'nunique'),
    hist_total_conditions=('condition_concept_id', 'count')
).reset_index()

# 3. Prediction Window (Target): Did they have high utilization AFTER the median date?
pred_df = df[df['condition_start_date'] > df['median_date']]
y_target = pred_df.groupby('person_id').size().reset_index(name='future_conditions')

# Merge features and target on person_id
model_df = X_features.merge(y_target, on='person_id', how='inner')

# Define a true prospective target: e.g., Top 25% future utilizers
utilization_threshold = model_df['future_conditions'].quantile(0.75)
model_df['target'] = (model_df['future_conditions'] > utilization_threshold).astype(int)

# 4. Prepare X, y, and groups for Patient-Level Splitting
X = model_df[['hist_distinct_conditions', 'hist_total_conditions']]
y = model_df['target']
groups = model_df['person_id']

print(f"Total valid patients for temporal prediction: {len(model_df)}")

# Patient-level train/test split
gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
train_idx, test_idx = next(gss.split(X, y, groups))

X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

# 5. Train Baseline Model
print("Training realistic prospective baseline model...")
baseline_model = LogisticRegression(class_weight='balanced', random_state=42)
baseline_model.fit(X_train, y_train)

# 6. Evaluate
y_pred_proba = baseline_model.predict_proba(X_test)[:, 1]
precision, recall, _ = precision_recall_curve(y_test, y_pred_proba)
pr_auc = auc(recall, precision)
brier = brier_score_loss(y_test, y_pred_proba)

print("\n--- Realistic Baseline Evaluation Results ---")
print(f"PR-AUC: {pr_auc:.4f}")
print(f"Brier Score (Calibration): {brier:.4f}")
```
